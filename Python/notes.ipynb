## Logical Operators
-   Evaluate multiple conditions `(or, and, not)`
-   `or` = at least one condition must be True
-   `and` = both conditions must be True
-   `not` = inverts the condition (not False, not True)

```python
# OR

temp = 25
is_raining = False

if temp > 35 or temp < 0 or is_raining:
    print("The outdoor event is cancelled")
else:
    print("The outdoor event is still happening")



```

```python
# AND

temp = 30
is_sunny = True

if temp >= 28 and is_sunny:
    print("It is HOT🥵")
elif temp <= 0 and is_sunny:
    print("Its COLD 🥶")
elif temp < 28 and temp > 0 and is_sunny:
    print("It is warm outside 😊")
    print("It is Sunny")
    

```

## Format Specifiers 

- `{value:flags}` format a value based on what flags are inserted
    - `.(numbers)f` = round to that many decimal places (fixed point)
    - `:(number)` = allocate that many times
    - `:<` = left justify
    - `:>` = right justify
    - `:^` = center justify
    - `:+` = use a plus sign to indicate positive value
    - `:` = insert a space before positive numbers
    - `:,` = comma separator

```python
price1 = 3.14159
price2 = -987.59
price3 = 12.34

print(f"Price 1 is ${price1:.2f}")
print(f"Price 2 is ${price2:.2f}")
print(f"Price 3 is ${price3:.2f}")

print(f"Price 1 is ${price1:<10}")
print(f"Price 2 is ${price2:<10}")
print(f"Price 3 is ${price3:<10}")
```

## While Loops
-  Execute some code WHILE some conditions remain true

```python
name = input("Enter your name")

while name == "":
    print("You did not enter your name")
    name = input("Enter your name")
else:
    print(f"Hello {name}")
```

```python
age = int(input("Enter your age: "))

while age < 0:
    print("Age cant be negative")
    age = int(input("Enter your age: "))
print(f"Your are {age} years old")
```

```python
food = input("Enter a food you like (q to quit)")
while food == "q":
    print(f"You like {food}")
    food = input("Enter a food you like (q to quit)")
print("bye")
```

```python
num = int(input("Enter a number between 1 - 10: "))

while num < 1 or num > 10:
    print(f"{num} is not valid")
    num = int(input("Enter a number between 1 - 10: "))
print(f"Your number is {num}")
```

## For loops

```python
import time 

my_time = int(input("Enter the time in seconds: "))
for x in range(my_time, 0, -1):
    seconds = x % 60
    minutes = int(x /60) % 60
    hours = int(x / 3600)
    print(f"{hours:02}:{minutes:02}:{seconds:02}")
    time.sleep(1)
print("Time's UP")
```

## Dictionary - 
- a collection of `{key:value}` pairs order and changeable. No duplicates

```python
capitals = {"USA": "Washington D.c",
            "India": "New Delhi",
            "China": "Beijing",
            "Russia": "Moscow"}

print(capitals.get("USA"))
print(capitals.get("India"))
print(capitals.get("Japan"))

if capitals.get("Japan"):
    print("That capital exist")
else:
    print("We have no knowledge of that capital")
```

```python
capitals.update({"Germany": "Berlin"})
capitals.update({"USA": "Detroit"})

print(capitals)
```

```python
capitals.pop("China")
```

## Return 
-  statement used to end a function
-  and send a result back to the caller

```python
def add(x, y):
    z = x + y
    return z

def subtract(x, y):
    z = x - y
    return z 

print(add(1,2))
print(subtract(1, 2))
```

```python
def create(first, last):
    first = first.capitalize()
    last = last.capitalize()
    return first + " " + last

full_name = create("Kai", "Kai")

print(full_name)
```
